```python
a = "  Hello World  "
#split
print(a)
b = a.split()
print(b)
#upper case
c =a.upper()
print(c)
#lower case
d = a.lower()
print(d)
#find
e = a.find('H')
print(e)
#replace
f = a.replace("World","Earth")
print(f)
#strip
g = a.strip()
print(g)
#count
A = a.count('o')
print(A)
#join
B = "-".join(b)
print(B)
```

```python
#program that count the no.of vowels and consonants in a string
a = "Lake"
b = a.lower()
vowels = "aeiou"
count_v = 0
count_c = 0
for ch in b:
  if ch.isalpha:
    if ch in vowels:
     count_v+=1
    else:
     count_c+=1
print("Total no.of vowels",count_v)
print("Total no.of consonants",count_c)
```
